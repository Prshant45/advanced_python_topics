```python

# Original DataFrame
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'], 
    'Age': [4, 5, 6],
    'Salary': [50000, 60000, None],
    'Department': [1, 2, 3],
    'Manager': [None, 'John', None],
    'skills': [['Python', 'SQL'], ['Java', 'C++'], None]
})

# New data repeating Department IDs 1, 2, and 1
new_data = pd.DataFrame({
    'Name': ['David', 'Emma', 'Frank'],
    'Age': [7, 8, 5],
    'Salary': [70000, 65000, 52000],
    'Department': [1, 2, 1],  # Repeated department IDs
    'Manager': ['Sarah', 'John', 'Sarah'],
    'skills': [['Go', 'Docker'], ['AWS', 'Python'], ['R']]
})

# Concatenate and reset index
df = pd.concat([df, new_data], ignore_index=True)
print(df)
```

```python
# substract
arr1 = np.array([[1, 2], [3, 4]], dtype=np.int32)
arr2 = np.array([[5, 6], [7, 8]], dtype=np.int32)
def add(a: np.int32, b: np.int32) -> np.int32:
    return a + b
# add - function name
# 2 - number of input arguments
# 1 - number of output arguments
my_add = np.frompyfunc(add, 2, 1 )
# my_add(arr1, arr2)
my_add(arr1, arr2)
```

```python
#np.vectorize(add)(arr1, arr2)
from numba import vectorize, float32, float64
@vectorize([float64(float32, float32)], target="parallel" )
def add(x, y):
    return x + y
arr1 = np.array([[1, 2], [3, 4]], dtype=np.float32)
arr2 = np.array([[5, 6], [7, 8]], dtype=np.float32)
add(arr1, arr2)
```
